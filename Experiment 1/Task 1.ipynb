#Creation of Numpy arrays of different Ranks or Dimensions

```python
import numpy as np
```

##1. np.array() function

```python
s1=np.array([1,2,3,4])
print(s1)
```

##2. Using np.zeros()/ np.ones()

###  2.1) np.zeros()

```python
a = np.zeros((3,5),dtype=np.uint64)
print(a)
```

###  2.2) np.ones()

```python
b = np.ones((3,3),dtype = np.uint8)
print(b)
```

##3. Using np.random.random()

```python
c = np.random.random((2,3))
print(c)
```

##4. Using np.arange()

```python
d = np.arange(10)
print(d)
```

##5. Using np.linespace()

```python
e = np.linspace(0.0,1.0,11)
print(e)
```

##6. Using np.copy()

```python
f = np.array([1,3,5,7])
x = f.copy()
print(x)
```

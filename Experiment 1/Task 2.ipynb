##Data Transformation
### Apply a custom function to a specific column of the dataset.

```python
def age_groups(age):
  if age<18:
    return 'Child'
  elif age<60:
    return 'Adult'
  else:
    return 'Senior'
```

```python
df['age_group'] = df['Age'].apply(age_groups)
df[['Age','age_group']].head(20)
```

```python
df.head()
```

###Map values of a column to new values using a dictionary.

```python
gender_mapping = {

        'male': 'M',
        'female':'F'

}
```

```python
df['Sex'] = df['Sex'].map(gender_mapping)
df[['Sex']].head()
```

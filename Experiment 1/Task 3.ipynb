#Find highest run getters in each of the seasons (2008-2022) and amount of his run. And plot a suitable chart indicating this information using matplotlib.

```python
df['Season'].unique()
```

```python
season_map = {
    '2020/21' : '2020',
    '2009/10': '2010',
    '2007/08': '2008'
}
```

```python
df['Season']=df['Season'].replace(season_map)
df['Season']=df['Season'].astype(int)
```

```python
df['Season']
```

```python
highest_scorers = (
    df.groupby(['batter','Season'])['batsman_run'].sum().reset_index()
)
```

```python
highest_scorers = highest_scorers.sort_values(['Season','batsman_run'],ascending=[True,False])
df = highest_scorers.groupby('Season').first().reset_index()
```

```python
df
```

```python
df.info()
```

```python
highest_scorers = (
    df.groupby(["Season", "batter"])["batsman_run"]
      .sum()
      .reset_index()
      .sort_values(["Season", "batsman_run"], ascending=[True, False])
      .groupby("Season")
      .first()
      .reset_index()
)
```

```python
highest_scorers.info()
```

```python
highest_scorers
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 10))

plt.pie(
    highest_scorers["batsman_run"],
    labels=highest_scorers["batter"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Highest Run Scorers in Each IPL Season (2008–2022)")
plt.show()
```

#Find out the top 5 batsman of the history of IPL based on strike rate. ( must faced at least 50 balls)

```python
df = df_ball.merge(df_match[["ID","Season"]], on="ID",how="left")
```

```python
df["extra_type"].value_counts(dropna=False)
```

```python
df[df["extra_type"] != "wides"]
```

```python
df[df["extra_type"].ne("wides") | df["extra_type"].isna()]
```

```python
batting = df[
    df["extra_type"].ne("wides") | df["extra_type"].isna()
].groupby("batter").agg(
    runs=("batsman_run", "sum"),
    balls_faced=("batter", "size")
).reset_index()

batting = batting[batting["balls_faced"] >= 50]

batting["strike_rate"] = (
    batting["runs"] / batting["balls_faced"]
) * 100

top_5 = batting.sort_values(
    "strike_rate", ascending=False
).head(5)

print(top_5)
```
